## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana tingkat pertumbuhan penyewaan sepeda bulanan (Month-over-Month) sepanjang tahun 2011 hingga 2012, dan pada bulan apa terjadi lonjakan penyewaan tertinggi?

Relevant: Sesuai dengan tujuan operasional bisnis peminjaman sepeda. Time-bound: Sepanjang tahun 2011-2012).*
  
- **Pertanyaan 2:** Berapa besar persentase penurunan rata-rata jumlah penyewaan sepeda saat cuaca buruk (Hujan/Salju Ringan) dibandingkan dengan cuaca cerah pada musim gugur (Fall) tahun 2012?

* **Pengecekan Keberadaan Data Kosong (*Missing Values*)**: Menjalankan fungsi `.isna().sum()` untuk mendeteksi apakah ada sel yang tidak terisi dalam dataset. Hasil pemeriksaan menunjukkan bahwa data ini bersih dari nilai kosong pada seluruh kolom.
* **Deteksi Data Duplikat**: Menggunakan fungsi `.duplicated().sum()` untuk menjamin bahwa setiap entri bersifat unik dan tidak ada rekaman ganda yang dapat membiaskan hasil statistik. Terdeteksi bahwa tidak ada baris data yang duplikat.
* **Analisis Nilai Kategorikal**: Melakukan inspeksi pada nilai-nilai yang di-*encode* seperti `season` dan `weathersit`. Meskipun secara teknis benar (integer), namun secara fungsional nilai-nilai ini sulit dipahami tanpa label teks, sehingga diidentifikasi sebagai masalah keterbacaan yang harus diperbaiki di tahap pembersihan.

Identifikasi ini sangat krusial agar kita bisa menyusun strategi pembersihan data yang tepat sasaran, sehingga analisis bisnis yang dihasilkan nantinya bersifat akurat dan reliabel.

### Cleaning Data

#### Fixing Data Quality Problem

Setelah mengidentifikasi beberapa isu pada tahap sebelumnya, saya melakukan langkah-langkah pembersihan (cleaning) untuk memastikan data siap dianalisis. Proses perbaikan ini meliputi:

*   **Konversi Tipe Data Tanggal**: Kita menggunakan fungsi `pd.to_datetime()` untuk mengubah kolom `dteday` dari tipe data *object* menjadi format *datetime*. Langkah ini sangat krusial agar Python dapat mengenali data tersebut sebagai urutan waktu, sehingga kita bisa melakukan analisis tren bulanan atau tahunan secara akurat.
*   **Transformasi Nilai Kategorikal (Mapping)**: Untuk meningkatkan keterbacaan data, kita melakukan pemetaan (*mapping*) pada nilai-nilai integer di kolom `season`, `yr`, dan `weathersit` menjadi label teks yang deskriptif. Misalnya, mengubah angka 1, 2, 3, 4 pada musim menjadi 'Spring', 'Summer', 'Fall', dan 'Winter'. Hal ini memudahkan kita (dan orang lain yang membaca laporan ini) dalam memahami informasi tanpa harus terus-menerus merujuk pada dokumentasi dataset.
*   **Feature Engineering (Penambahan Kolom)**: Kita juga menambahkan kolom baru bernama `month_year` dengan mengekstraksi informasi bulan dan tahun dari kolom tanggal. Penambahan ini bertujuan untuk mempermudah proses pengelompokan (grouping) saat kita ingin menjawab pertanyaan bisnis mengenai tren pertumbuhan penyewaan secara bulanan.

Dengan melakukan langkah-langkah di atas, dataset kini telah bertransformasi dari data mentah yang sulit dibaca menjadi dataset yang bersih, informatif, dan siap untuk masuk ke tahap analisis eksploratif.

```python
day_df['dteday'] = pd.to_datetime(day_df['dteday'])
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])

season_mapping = {1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
year_mapping = {0: '2011', 1: '2012'}
weather_mapping = {1: 'Clear/Partly Cloudy', 2: 'Mist/Cloudy', 3: 'Light Snow/Rain', 4: 'Heavy Rain/Snow'}

day_df['season_label'] = day_df['season'].map(season_mapping)
day_df['year_label'] = day_df['yr'].map(year_mapping)
day_df['weather_label'] = day_df['weathersit'].map(weather_mapping)

day_df['month_year'] = day_df['dteday'].dt.to_period('M')

print("Tipe data setelah diperbaiki:")
print(day_df.dtypes[['dteday', 'season_label', 'year_label', 'weather_label']])
```

**Insight:** (Opsional)
- Tipe data pada kolom `dteday` telah berhasil dikonversi dari `object` (string) menjadi `datetime64[ns]`, sehingga fitur-fitur berbasis waktu dapat diekstraksi dan dianalisis dengan lebih mudah.
- Transformasi data kategorikal (seperti musim, tahun, dan cuaca) dari format angka menjadi label teks representatif (bertipe `object`) telah berhasil dilakukan. Hal ini akan sangat membantu dalam memperjelas interpretasi saat proses Exploratory Data Analysis (EDA) dan pembuatan legenda pada visualisasi data.

## Exploratory Data Analysis (EDA)

### Explore Data Bike Sharing

Pada tahap Exploratory Data Analysis (EDA), kita melakukan penggalian informasi lebih dalam untuk memahami pola, hubungan, dan tren yang tersembunyi di dalam dataset. Langkah-langkah eksplorasi yang dilakukan meliputi:

*   **Analisis Tren Waktu (Time-Series Exploration)**: Kita melakukan agregasi data jumlah penyewaan sepeda berdasarkan urutan bulan dan tahun. Proses ini bertujuan untuk melihat gambaran besar performa bisnis dari waktu ke waktu, sehingga kita dapat mengidentifikasi kapan periode puncak (peak season) dan periode lesu terjadi.
*   **Analisis Pengaruh Cuaca**: Kita mengeksplorasi bagaimana variabel kondisi cuaca memengaruhi rata-rata penyewaan harian. Dengan memfokuskan observasi pada Musim Gugur (Fall) tahun 2012, kita dapat mengukur seberapa sensitif pengguna terhadap perubahan cuaca, yang sangat krusial untuk pengambilan keputusan operasional.
*   **Analisis Perilaku Pengguna (User Type Analysis)**: Kita membedah total penyewaan menjadi dua kategori: pengguna 'Casual' dan 'Registered'. Eksplorasi ini membantu kita memahami profil pelanggan kita; apakah mayoritas pengguna adalah pelanggan setia yang menggunakannya untuk kebutuhan harian (seperti bekerja), atau pengguna kasual yang hanya menggunakannya untuk rekreasi.
*   **Ringkasan Statistik Dasar**: Menggunakan fungsi statistik untuk melihat distribusi data, nilai rata-rata, serta nilai minimum dan maksimum dari jumlah penyewaan. Hal ini memberikan fondasi angka yang kuat sebelum kita menerjemahkannya ke dalam bentuk visualisasi.

Melalui proses eksplorasi ini, kita tidak hanya sekadar melihat angka, tetapi juga mendapatkan "cerita" di balik data yang akan menjadi dasar kuat dalam menjawab pertanyaan bisnis yang telah ditentukan.

```python
monthly_rentals = day_df.groupby('month_year')['cnt'].sum().reset_index()
monthly_rentals['month_year'] = monthly_rentals['month_year'].astype(str)

fall_2012_df = day_df[(day_df['season_label'] == 'Fall') & (day_df['year_label'] == '2012')]
weather_impact = fall_2012_df.groupby('weather_label')['cnt'].mean().reset_index().sort_values(by='cnt', ascending=False)

display(monthly_rentals.head())
display(weather_impact)
```

**Insight:** (Opsional)
- Dari hasil eksplorasi tren penyewaan bulanan (`monthly_rentals`), terlihat adanya fluktuasi jumlah peminjaman yang membentuk pola musiman. Jumlah penyewaan cenderung selalu rendah di awal dan akhir tahun, namun konsisten merangkak naik hingga mencapai puncaknya pada periode kuartal ketiga (sekitar bulan Agustus hingga September). Secara keseluruhan, tahun 2012 juga menunjukkan volume penyewaan yang lebih tinggi dibandingkan tahun 2011.
- Berdasarkan hasil agregasi rata-rata penyewaan di Musim Gugur tahun 2012 (`weather_impact`), didapatkan informasi bahwa kondisi cuaca sangat berkorelasi dengan minat pengguna untuk menyewa sepeda. Cuaca yang cerah atau sedikit berawan (`Clear/Partly Cloudy`) mencatatkan angka rata-rata penyewaan harian tertinggi, sedangkan cuaca buruk (`Light Snow/Rain`) menunjukkan angka rata-rata penyewaan yang paling rendah.

## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
plt.figure(figsize=(16, 6))

sns.lineplot(
    data=monthly_rentals,
    x='month_year',
    y='cnt',
    marker='o',
    linewidth=2.5,
    color='#008080' # Teal
)

plt.title('Tren Total Penyewaan Sepeda Bulanan (2011 - 2012)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Bulan-Tahun', fontsize=12)
plt.ylabel('Total Penyewaan (cnt)', fontsize=12)
plt.xticks(rotation=45)

peak_month = monthly_rentals.loc[monthly_rentals['cnt'].idxmax()]
plt.scatter(peak_month['month_year'], peak_month['cnt'], color='#E67E22', s=150, zorder=5) # Orange

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Pertanyaan 2:

```python
plt.figure(figsize=(10, 6))

custom_palette = {'Clear/Partly Cloudy': '#008080', 'Mist/Cloudy': '#73A8A8', 'Light Snow/Rain': '#E67E22'}

sns.barplot(
    data=weather_impact,
    x='weather_label',
    y='cnt',
    palette=custom_palette,
    hue='weather_label',
    legend=False
)

plt.title('Rata-rata Penyewaan Sepeda Berdasarkan Kondisi Cuaca\n(Musim Gugur 2012)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Kondisi Cuaca', fontsize=12)
plt.ylabel('Rata-rata Penyewaan Harian', fontsize=12)

for index, row in enumerate(weather_impact['cnt']):
    plt.text(index, row + 100, f'{round(row)}', color='black', ha="center", fontsize=11)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

Titik puncak (*peak*) penyewaan tertinggi sepanjang sejarah data terjadi pada bulan September 2012.

Pengguna jauh lebih menyukai bersepeda saat cuaca cerah atau sedikit berawan (*Clear/Partly Cloudy*) dengan rata-rata tembus 7.359 penyewaan per hari. Ketika cuaca memburuk menjadi hujan/salju ringan (*Light Snow/Rain*), angka tersebut merosot tajam menjadi hanya 5.239 penyewaan per hari.

## Analisis Lanjutan (Opsional) Clustering Waktu Operasional

```python
def get_time_category(hr):
    if 5 <= hr <= 11:
        return '1. Morning (05:00 - 11:00)'
    elif 12 <= hr <= 16:
        return '2. Afternoon (12:00 - 16:00)'
    elif 17 <= hr <= 20:
        return '3. Evening (17:00 - 20:00)'
    else:
        return '4. Night (21:00 - 04:00)'

hour_df['time_category'] = hour_df['hr'].apply(get_time_category)

user_type_time = hour_df.groupby('time_category')[['casual', 'registered']].mean().reset_index()

user_type_melted = pd.melt(
    user_type_time,
    id_vars=['time_category'],
    value_vars=['casual', 'registered'],
    var_name='user_type',
    value_name='avg_rentals'
)

display(user_type_time)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=user_type_melted,
    x='time_category',
    y='avg_rentals',
    hue='user_type',
    palette={'registered': '#008080', 'casual': '#E67E22'}
)

plt.title('Clustering Perilaku Pengguna: Casual vs Registered Berdasarkan Waktu', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Kategori Waktu (Clustered)', fontsize=12)
plt.ylabel('Rata-rata Penyewaan', fontsize=12)
plt.legend(title='Tipe Pengguna', loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
